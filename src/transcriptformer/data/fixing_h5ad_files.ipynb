```python
import anndata as ad
import mygene
from pathlib import Path

base = Path.home() / "Desktop" / "Transcriptformer"
adata = ad.read_h5ad(base / "heart_subset.h5ad")

gene_symbols = adata.var_names.tolist()

mg = mygene.MyGeneInfo()

results = mg.querymany(
    gene_symbols,
    scopes="symbol",
    fields="flybase",
    species="7227",
    as_dataframe=True
)

print(results.head())
print(results.columns)
```

```python
results = mg.querymany(
    gene_symbols[:20],
    scopes="symbol",
    fields="all",
    species="7227",
    as_dataframe=True
)

print(results.columns.to_list())
print(results.head())
```

```python
import anndata as ad
from pathlib import Path

base = Path.home() / "Desktop" / "Transcriptformer"
adata = ad.read_h5ad(base / "heart_subset.h5ad")

mapping = results["FLYBASE"].to_dict()

adata.var["ensembl_id"] = adata.var_names.map(mapping)

print(adata.var["ensembl_id"].head(20).to_list())
print("FBgn count:", adata.var["ensembl_id"].astype(str).str.startswith("FBgn").sum())
print("Missing:", adata.var["ensembl_id"].isna().sum())

adata = adata[:, adata.var["ensembl_id"].notna()].copy()
adata.write_h5ad(base / "heart_subset_fixed.h5ad")

print("Saved fixed file.")
```

```python
import anndata as ad
import mygene
from pathlib import Path

base = Path.home() / "Desktop" / "Transcriptformer"
adata = ad.read_h5ad(base / "heart_subset.h5ad")

gene_symbols = adata.var_names.tolist()

mg = mygene.MyGeneInfo()

results = mg.querymany(
    gene_symbols,
    scopes="symbol",
    fields="FLYBASE",
    species="7227",
    as_dataframe=True
)

mapping = results["FLYBASE"].to_dict()

adata.var["ensembl_id"] = adata.var_names.map(mapping)

print("FBgn count:", adata.var["ensembl_id"].astype(str).str.startswith("FBgn").sum())
print("Missing:", adata.var["ensembl_id"].isna().sum())

adata = adata[:, adata.var["ensembl_id"].notna()].copy()
adata.write_h5ad(base / "heart_subset_fixed.h5ad")

print("Saved:", base / "heart_subset_fixed.h5ad")
```
